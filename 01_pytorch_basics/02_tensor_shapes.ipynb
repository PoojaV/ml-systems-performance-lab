```python
import torch

# Scalar
a = torch.tensor(5)

# Vector
b = torch.tensor([1, 2, 3])

# Matrix
c = torch.tensor([
    [1, 2],
    [3, 4]
])

print(a.shape)
print(b.shape)
print(c.shape)
```

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)
```
